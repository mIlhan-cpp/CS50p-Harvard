```python
import requests
import json
sys.argv=["script.py",'weezer']
if len(sys.argv)!=2:
    sys.exit()

resp=requests.get("https://itunes.apple.com/search?entity=song&limit=1&term=weezer" + sys.argv[1])
print(json.dumps(resp.json(),indent=2))
```
